```python
import pandas as pd
df = pd.read_csv(r"C:\Users\yuraz\Downloads\russian_novels_metadata.csv")
df.shape
```

```python
df.info()
```

```python
import matplotlib.pyplot as plt
a = (df[df['author_gender']=='женский'])
b = a['general_genre'].value_counts()
plt.bar(b.index, b.values)
plt.title('Распределение для женщин')
plt.xlabel("Жанр")
plt.ylabel("Количество")
plt.xticks(rotation=90)
plt.show()
print(b[:3])
```
